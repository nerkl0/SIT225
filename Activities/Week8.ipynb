```python
import csv
import pandas as pd
import matplotlib.pyplot as plt

acc_X_df = pd.read_csv('./data/accelerometer_X.csv')
acc_Y_df = pd.read_csv('./data/accelerometer_Y.csv')
acc_Z_df = pd.read_csv('./data/accelerometer_Z.csv')

# plot X accelerometer
acc_X_df.plot(figsize=(10,5))
```

```python

dfs = [acc_X_df, acc_Y_df, acc_Z_df]
labels = ['X', 'Y', 'Z']

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

for ax, df, label in zip(axes, dfs, labels):
    ax.plot(df['value'])
    ax.set_title(f'Accelerometer {label}')
    ax.set_xlabel('Sample')

axes[0].set_ylabel('Acceleration')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20, 6))
for df, label in zip(dfs, labels):
    plt.plot(df['value'], label=label, alpha=0.7, linewidth=0.8)
plt.xlabel('Sample')
plt.ylabel('Acceleration')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(20, 6))
for df, label in zip(dfs, labels):
    plt.plot(df['value'].rolling(50).mean(), label=label, linewidth=1)
plt.xlabel('Sample')
plt.ylabel('Acceleration')
plt.legend()
plt.show()
    
    
```
